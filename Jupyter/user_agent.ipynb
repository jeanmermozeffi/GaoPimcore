```python
import httpagentparser
import requests
from random import randint
```

```python
def is_valid_user_agent(user_agent: str) -> bool:
    try:
        parsed_agent = httpagentparser.detect(user_agent)
        # Si le user-agent est valide, il doit contenir des informations sur le navigateur ou le système d'exploitation
        if parsed_agent.get('browser') or parsed_agent.get('platform'):
            return True
    except Exception as e:
        print(f"Erreur lors de l'analyse du user-agent: {e}")
    return False
```

```python
user_agent_test = "Mozilla/5.0 (Wnehkrfgkuwjfbkuw"

if is_valid_user_agent(user_agent_test):
    print("User-agent valide")
else:
    print("User-agent invalide")
```

```python
SCRAPEOPS_API_KEY = '1cdf4aa4-55a5-4534-bfcf-aa6872a9e7da'
```

```python
def get_headers_list():
    response = requests.get('http://headers.scrapeops.io/v1/browser-headers?api_key=' + SCRAPEOPS_API_KEY)
    json_response = response.json()
    return json_response.get('result', [])

def get_user_agent_list():
    response = requests.get('http://headers.scrapeops.io/v1/user-agents?api_key=' + SCRAPEOPS_API_KEY)
    json_response = response.json()
    return json_response.get('result', [])

def get_random_header(header_list):
    random_index = randint(0, len(header_list) - 1)
    return header_list[random_index]
```
